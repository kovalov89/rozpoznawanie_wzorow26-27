```python

import numpy as np

rng = np.random.default_rng(42)

num_clusters = 3
num_features = 16
k = 5

data = []
labels = []
seeds = []
probabilities = []

for cluster in range(num_clusters):
    seed = rng.integers(0, 2, size=num_features, dtype=np.uint8)
    p = 0.1 * np.sum(seed) / num_features

    seeds.append(seed)
    probabilities.append(p)

    data.append(seed.copy())
    labels.append(cluster + 1)

    for _ in range(k):
        flip_mask = (rng.random(num_features) < p).astype(np.uint8)
        sample = np.bitwise_xor(seed, flip_mask)

        data.append(sample)
        labels.append(cluster + 1)

X = np.array(data, dtype=np.uint8)

print("Liczba obiektów:", len(X))
print("Liczba cech:", X.shape[1])

print("\nWektory bazowe:")
for i, seed in enumerate(seeds):
    print(
        f"Klaster {i+1}: {''.join(map(str, seed))}, "
        f"p = {probabilities[i]:.5f}"
    )

print("\nWygenerowany zbiór:")
for i, row in enumerate(X):
    vector = ''.join(map(str, row))
    print(f"Obiekt {i+1:02d} | Klaster {labels[i]} | {vector}")
```
